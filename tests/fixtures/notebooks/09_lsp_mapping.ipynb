# LSP mapping

Markdown between code cells must not shift diagnostics incorrectly.


```python
from dataclasses import dataclass

@dataclass
class Точка:
    x: float
    y: float

def length(point: Точка) -> float:
    return (point.x ** 2 + point.y ** 2) ** 0.5
```

```python
%matplotlib inline
!echo ignored-by-lsp
point = Точка(3, 4)
```

## Result


```python
length(point)
```
